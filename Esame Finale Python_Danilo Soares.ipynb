## 2. Per continente

### a. Numero di casi sin dal inizio della pandemia:

```python
# filtando i paesi dalle vaoci globali ed aggregati

df_c = df[df["continent"].notna()]
df_c
```

```python
# Ordina per location e data, poi prendi l'ultima riga per ciascun paese
df_c["date"] = pd.to_datetime(df_c["date"])
df_ultimi = (
    df_c.sort_values(["location", "date"])
        .groupby("location")
        .tail(1)
)

# Controllo veloce
df_ultimi[["location", "continent", "date", "total_cases"]].head()
```

```python
# Somma dei casi per continente
casi_cont = df_ultimi.groupby("continent")["total_cases"].sum()

# Totale mondiale (somma di tutti i paesi)
totale_mondiale = casi_cont.sum()

print("Casi totali per continente:")
print(casi_cont)
```

### b. Percentuale rispetto al totale mondiale del numero di casi

```python
# Percentuali per continente

perc_cont = casi_cont / totale_mondiale * 100

print("Percentuale sul totale mondiale:")
print(perc_cont)
```

## 3. Nuovi casi settimanali in italia enl 2022 al netto dei giorni senza misurazione e grafici riassuntivi
### a. Evoluzione dei casi totali dal inizio alla fine del anno in Italia

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_it_2022 = df[
    (df["location"] == "Italy") &
    (df["date"].between("2022-01-01", "2022-12-31"))
]

# Tieni solo i giorni con valore > 0 (no NaN, no zeri)
df_it_2022 = df_it_2022[df_it_2022["new_cases"] > 0]

df_it_2022["date"] = pd.to_datetime(df_it_2022["date"])

df_it_2022
```

```python
import seaborn as sns

df_it_2022["date"] = pd.to_datetime(df_it_2022["date"])

# serie settimanale
casi_settimane_22 = (
    df_it_2022
    .set_index("date")["new_cases"]
    .resample("W")
    .sum()
    .reset_index()
)

casi_settimane_22.plot(x="date", y="new_cases")
```

## 4. Italia, Germani e Grancia

### a. Pazienti in terapia intensiva

```python
countries = ["Italy","Germany","France"]
mask = (
    df["location"].isin(countries) &
    (df["date"] >= "2022-05-01") &
    (df["date"] <= "2023-04-30")
)
df_icu = df.loc[mask, ["location","date","icu_patients"]].dropna()

plt.figure(figsize=(6,4))
sns.boxplot(data=df_icu, x="location", y="icu_patients")
plt.tight_layout()
plt.show()
```

### b. Commento
    Germania e Francia per il periodo considerato mostrano valori mediani piu elevati di pazienti in terapia intensiva ed una dispersione elevata, L'Italia per sua volta presenta una dispersione piu contenuta e livelli nettamente inferiori di pazienti in terapia intensiva. I dati permetono la lettura di una maggiore pressione sulle terapia intensive in Francia e Germania rispetto all'Italia.

## 5. Italia, Germania, Francia e Spagna 2021

### a. La somma dei pazienti ospitalizzati

```python
paesi = ["Italy","Germany","France","Spain"]
mask = (
    df["location"].isin(paesi) &
    (df["date"] >= "2021-01-01") &
    (df["date"] <= "2021-12-31")
)
df_hosp = df.loc[mask, ["location","date","hosp_patients"]].dropna()

hosp_sum = df_hosp.groupby("location")["hosp_patients"].sum()
print(hosp_sum)
```

```python
df_hosp_bar = hosp_sum.reset_index()
df_hosp_bar.columns = ["location","hosp_patients_sum"]

plt.figure(figsize=(6,4))
sns.barplot(data=df_hosp_bar, x="location", y="hosp_patients_sum")
plt.tight_layout()
plt.show()
```

### B. La gestione dei nulli senza perdita di informazione rilenvante
    Ci sono molti dati nulli probabilmente perche non tutti i paesi riportano i dati sui ricoveri in modo continuo. Sostituendo con "0" indicando nessun registro in quei giorni. Il rischio sarebbe quello di sottostimare  la somma nel raport finale, pero essendo indicativo il confronto tra i paesi potrebbe non essere un rischio rilevante potendo sempre nel caso escludere i nulli dei gruppi.
